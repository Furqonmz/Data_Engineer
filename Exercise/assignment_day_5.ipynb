Part A - Data Structure
Lakukan eksplorasi terhadap struktur data berikut:

List: Buat list nama buah, tambahkan satu buah baru, lalu tampilkan panjang list.
Tuple: Buat tuple berisi tiga hingga lima kota, tampilkan kota kedua.
Set: Buat set angka ganjil dari 1–10, tambahkan angka baru, lalu cetak isi set.
Dictionary: Buat dictionary data siswa (nama, umur, jurusan) dan akses nilai jurusan.

```python
# List
buah = ["apel", "jeruk", "pisang", "mangga"]
buah.append("salak")
print(f"Daftar Buah     : {buah}")
print(f"Panjang list    : {len(buah)}")

# Tuple
city = ("Bandung", "Jakarta", "Surabaya", "Semarang", "Yogyakarta")
print(f"Kota Kedua      : {city[1]}")

# Set
angka = {1, 3, 5, 7, 9}
angka.add(11)
print(f"Isi dari set    : {angka}")

# Dictionary
data_siswa = {
    "nama" : "Furqon",
    "umur" : 23,
    "jurusan" : "Teknik Elektro"
}
print(f"Jurusan         : {data_siswa["jurusan"]}")
```

Part B – Pernyataan Kondisional if elif else
Gunakan konsep if, elif, dan else untuk menyelesaikan soal berikut:

Buat variabel nilai.
Jika nilai ≥ 75, tampilkan "Lulus".
Jika nilai antara 60–75, tampilkan "Remedial".
Jika nilai < 60, tampilkan "Tidak lulus".

```python
nilai = 60

if nilai >= 75:
    print("Lulus")
elif nilai >= 60:
    print("Remedial")
else:
    print("Tidak Lulus")
```

Part C – Pengulangan / Loops
Gunakan for loop dan while loop untuk menyelesaikan soal berikut:

Tampilkan angka 1 sampai 10 menggunakan for.
Tampilkan semua elemen dari list berikut:
buah = ["apel", "jeruk", "pisang", "mangga"]

```python
for i in range(1, 11):
    print(f"{i}")

print("\n")

buah = ["apel", "jeruk", "pisang", "mangga"]
for b in buah:
    print(b)

print("\n")
# Perulangan menggunakan while
x = 1
while x <= 10:
    print(f"{x}")
    x += 1
```

Part D – Fungsi
Gunakan konsep function untuk menyelesaikan soal berikut:

Buat fungsi volume_balok(panjang, lebar, tinggi)
Kembalikan hasil volumenya dan tampilkan hasilnya.

```python
def volume_balok(panjang, lebar, tinggi):
    return panjang * lebar * tinggi

print(f"Volume balok    : {volume_balok(3,4,5)} cm³")
```

Part E – Project
Buatlah program kasir sederhana, dengan aturan:

User dapat menginput nama, harga dan jumlah barang (gunakan fungsi input)
User dapat menginput berulang-ulang (gunakan pengulangan for)
Output berupa nama-nama barang yang diinput dan total harga semua barang, dipotong harga diskon sebagai berikut:
Belanja >= 100.000 diskon 10%
Belanja >= 50.000 diskon 5%
Program ini dapat diulang jika user ingin menghitung barang baru (gunakan pengulangan while)

```python
print("Memulai Transaksi [y/n]\n")
x = str(input("Mulai Transaksi?").lower())

while x != "n":
    jumlah_barang = int(input("Berapa jenis yang dibeli"))
    total_belanja = 0

    print("================TOKO KUE CATREST================")
    print("------------------------------------------------")
    print(f"{'Item':<18}{'Jumlah':>5}{'Harga':>10}{'Subtotal':>15}")
    print("------------------------------------------------")

    for i in range(jumlah_barang):
        nama_barang = input("Nama Barang? ")
        harga_barang = int(input("Harga barang "))
        jumlah = int(input("Jumlah barang yang dibeli"))

        sub_total = harga_barang * jumlah
        total_belanja += sub_total

        print(f"{nama_barang:<18}{jumlah:>5}{'Rp ' + format(harga_barang, ','):>10}{'Rp ' + format(sub_total, ','):>15}")

    if total_belanja >= 100000:
        disc = 0.1*total_belanja
    elif total_belanja >=50000:
        disc = 0.05*total_belanja
    else:
        disc = 0

    total_bayar = total_belanja - disc

    print("------------------------------------------------")
    print(f"{'Total Belanja':<25}{': Rp ' + format(total_belanja, ',.0f'):>23}")
    print(f"{'Total Diskon':<25}{': Rp ' + format(disc, ',.0f'):>23}")
    print(f"{'Total Bayar':<25}{': Rp ' + format(total_bayar, ',.0f'):>23}")
    print(f"\nApakah ingin memulai transaksi kembali? [y/n]\n")
    x = str(input("Ulangi Pembayaran").lower())
```
