PERSONAL EXPENSE TRACKER

```python
expense_tracker = []
x = str(input().lower())

while x != "n":
    pengeluaran = {}
    pengeluaran["nama"] = str(input("nama pengeluaran"))
    pengeluaran["kategori"] = str(input("Kategori"))
    pengeluaran["harga"] = int(input("Harga"))
    print(f"\n{pengeluaran}\n")

    expense_tracker.append(pengeluaran)
    x = str(input().lower())

print(expense_tracker)
```
